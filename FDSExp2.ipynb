```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# -------------------------------------------------
# 1. CREATE DATASET 
#-------------------------------------------------
data = {
    "Date":[
        "01-01-2023", "02-01-2023", "03-01-2023", "04-01-2023",
        "05-01-2023", "06-01-2023", "07-01-2023", "08-01-2023",
        "09-01-2023", "10-01-2023", "11-01-2023", "12-01-2023",
        "13-01-2023", "14-01-2023", "15-01-2023", "16-01-2023"
    ],
    "Product": [
        "Product A", "Product B", "Product A", "Product C",
        "Product B", "Product A", "Product C", "Product B",
        "Product A", "Product C", "Product B", "Product A",
        "Product C", "Product B", "Product A", "Product C"
    ],
    "Sales": [
        200, 150, 220, 300, 180, 210, 320, 160,
        230, 310, 190, 240, 330, 170, 250, 340
    ],
    "Quantity": [
        4, 3, 5, 6, 4, 5, 7, 3,
        6, 7, 4, 6, 8, 3, 7, 8
    ],
    "Region": [
        "North", "South", "North", "East",
        "West", "North", "East", "South",
        "North", "East", "West", "North",
        "East", "South", "North", "East"
    ]
}
# Create DataFrame
df = pd.DataFrame(data)
# Convert Date column
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)
# -------------------------------------------------
# 2. DISPLAY DATA
# -------------------------------------------------
print("DATASET:")
print(df.head())
print("\n")
# -------------------------------------------------
# 3. CHECK MISSING VALUES
# -------------------------------------------------
print("MISSING VALUES:")
print(df.isnull().sum())
print("\n")
# -------------------------------------------------
# 4. STATISTICAL SUMMARY
# -------------------------------------------------
print("STATISTICAL SUMMARY:")
print(df[["Sales", "Quantity"]].describe())
print("\n")
# -------------------------------------------------
# 5. PRODUCT-WISE SALES AND QUANTITY
# -------------------------------------------------
product_summary = df.groupby("Product")[["Sales", "Quantity"]].sum().reset_index()
print("PRODUCT-WISE SALES AND QUANTITY:")
print(product_summary)
print("\n")
# -------------------------------------------------
# 6. LINE GRAPH - SALES OVER DATE
# -------------------------------------------------
plt.figure(figsize=(10, 5))
plt.plot(
    df["Date"],
    df["Sales"],
    marker="o",
    color="blue",
    linewidth=2
)
plt.title("Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
# -------------------------------------------------
# 7. BAR GRAPH - PRODUCT-WISE SALES
# -------------------------------------------------
plt.figure(figsize=(8, 5))
plt.bar(
    product_summary["Product"],
    product_summary["Sales"],
    color=["blue", "orange", "green"]
)
plt.title("Product-wise Total Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
# -------------------------------------------------
# 8. CORRELATION MATRIX
# -------------------------------------------------
correlation = df[["Sales", "Quantity"]].corr()
print("CORRELATION MATRIX:")
print(correlation)
print("\n")
# -------------------------------------------------
# 9. CORRELATION HEATMAP
# -------------------------------------------------
plt.figure(figsize=(7, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5,
    fmt=".2f"
)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```
