```python
import numpy as np
import pandas as pd

df = pd.read_csv("Employee_Dataset.csv")

print("ORIGINAL DATASET")
print(df)

print("\nDATASET INFORMATION")
df.info()

print("\nDEPARTMENT MODE")
print(df.Department.mode())

df["Department"] = df["Department"].fillna(df["Department"].mode()[0])

df["Age"] = df["Age"].fillna(df["Age"].median())

df["Salary"] = df["Salary"].fillna(round(df["Salary"].mean()))

print("\nDATASET AFTER FILLING MISSING VALUES")
print(df)

print("\nDUMMY VALUES")
print(pd.get_dummies(df.Department))

updated_dataset = pd.concat(
    [pd.get_dummies(df.Department), df.iloc[:, [1, 2, 3]]],
    axis=1
)

print("\nUPDATED DATASET")
print(updated_dataset)

updated_dataset["Promoted"] = updated_dataset["Promoted"].replace(
    ["No", "Yes"], [0, 1]
)

print("\nFINAL DATASET")
print(updated_dataset)
```
