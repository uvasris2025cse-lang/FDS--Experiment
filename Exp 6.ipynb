```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Country": ["India", "Japan", "China", "India", "Japan",
                "China", "India", "China", "Japan", "India"],
    "Age": [22, 31, 28, 35, 42, 26, 39, 45, 30, 34],
    "Salary": [35000, 52000, 46000, 61000, 75000,
               41000, 68000, 82000, 50000, 57000]
})

print("Original Dataset:")
print(df)

country = pd.get_dummies(df["Country"], dtype=int)

data = pd.concat([country, df[["Age", "Salary"]]], axis=1)

print("\nPrepared Dataset:")
print(data)

mean = data.mean()
std = data.std()

standard_scaled = (data - mean) / std

print("\nStandard Scaled Data:")
print(standard_scaled)

minimum = data.min()
maximum = data.max()

minmax_scaled = (data - minimum) / (maximum - minimum)

print("\nMin-Max Scaled Data:")
print(minmax_scaled)
```
