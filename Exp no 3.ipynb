```python
import numpy as np
import pandas as pd

df = pd.read_csv("Student_Dataset.csv")

print("ORIGINAL DATASET")
print(df)

print("\nDUPLICATE ROWS")
print(df.duplicated())

df.drop_duplicates(inplace=True)

index = np.array(list(range(len(df))))
df.set_index(index, inplace=True)

print("\nDATASET AFTER REMOVING DUPLICATES")
print(df)

df.loc[(df["Mark"] < 0) | (df["Mark"] > 100), "Mark"] = np.nan
df.loc[(df["Attendance"] < 0) | (df["Attendance"] > 100), "Attendance"] = np.nan
df.loc[df["Salary"] < 0, "Salary"] = np.nan

print("\nDATASET AFTER REPLACING INVALID VALUES")
print(df)

print("\nUNIQUE DEPARTMENTS")
print(df["Department"].unique())

print("\nUNIQUE PREFERENCES")
print(df["Preference"].unique())

df["Preference"] = df["Preference"].replace("coding", "Coding")
df["Preference"] = df["Preference"].replace("music", "Music")

print("\nDATASET AFTER CORRECTING VALUES")
print(df)

df["Salary"] = df["Salary"].fillna(round(df["Salary"].mean()))
df["Mark"] = df["Mark"].fillna(round(df["Mark"].median()))
df["Attendance"] = df["Attendance"].fillna(round(df["Attendance"].median()))
df["Preference"] = df["Preference"].fillna(df["Preference"].mode()[0])

print("\nFINAL CLEANED DATASET")
print(df)
```
